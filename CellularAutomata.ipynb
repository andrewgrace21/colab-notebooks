```python
import matplotlib.pyplot as plt
import random

TIMESTEPS = 50
START = [random.randint(0,1) for _ in range(200)]
LENGTH = 2*TIMESTEPS + len(START)
print(START)

def binary(num,digits):
    b = [(num//(2**i))%2 for i in range(digits)]
    return b

for r in range(2**8):
    print(r)
    display = [[0]*TIMESTEPS + START + [0]*TIMESTEPS]
    R = binary(r,8)
    RULE = {i : R[i] for i in range(len(R))}
    for t in range(1,TIMESTEPS):
        line = []
        for l in range(LENGTH):
            if(l==0):
                line.append(RULE[2*display[t-1][l]+display[t-1][l+1]])
            elif(l==LENGTH-1):
                line.append(RULE[4*display[t-1][l-1]+2*display[t-1][l]])
            else:
                line.append(RULE[4*display[t-1][l-1]+2*display[t-1][l]+display[t-1][l+1]])
        display.append(line)

    fig, ax = plt.subplots()
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlim(TIMESTEPS/2,LENGTH-TIMESTEPS/2)
    im = ax.imshow(display, cmap='binary')
    plt.show()
```
