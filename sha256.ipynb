```python
# Algorithm Implementation: https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf
# Youtube Tutorial: https://www.youtube.com/watch?v=orIgy2MjqrA

import math
import numpy as np
from sympy import prime

#USER_STRING = input('Encrypt: ')
USER_STRING = 'RedBlockBlue'

def main():
    ASCII = ''.join([str(binary(ord(list(USER_STRING)[i]),8)) for i in range(len(USER_STRING))])
    padded = str(ASCII)+'1'+'0'*(math.ceil((len(ASCII)+65)/512)*512-len(ASCII)-65)+str(binary(len(ASCII),64))
    H = [binary(math.sqrt(prime(i+1))%1,-32) for i in range(8)]
    K = [binary(np.cbrt(prime(i+1))%1,-32) for i in range(64)]
    M = [[str(padded[i+j:i+j+32]) for j in range(0,512,32)] for i in range(0,len(padded),512)]
    for N in range(int(len(padded)/512)):
        W = M[N]
        for t in range(16,64):
            W.append(binary_add([sigma_one(W[t-2]),W[t-7],sigma_zero(W[t-15]),W[t-16]],32))
        H = compute_intrm(H,K,W)
    print(f"{int(''.join(H),2):x}")

def binary(num,digits):
    b = [str((num//(2**i))%2) for i in range(digits)]
    if(digits<0):
        b = [str(int((num//(float(2)**(-i)))%2)) for i in list(reversed(range(1,abs(digits)+1)))]
    return ''.join(b[::-1])

def binary_add(nums,digits):
    sum = int(nums[0],2)
    for i in range(1,len(nums)):
        sum += int(nums[i],2)
    return binary(sum,digits)

def sigma_zero(num):
    ROTR_seven = num[-7:]+num[:-7]
    print(ROTR_seven)
    ROTR_eighteen = num[-18:]+num[:-18]
    print(ROTR_eighteen)
    SHR_three = ('0'*3)+num[:-3]
    print(SHR_three)
    return ''.join([str((int(ROTR_seven[i])+int(ROTR_eighteen[i])+int(SHR_three[i]))%2) for i in range(len(num))])

def sigma_one(num):
    ROTR_seventeen = num[-17:]+num[:-17]
    ROTR_eighteen = num[-18:]+num[:-18]
    SHR_ten = ('0'*10)+num[:-10]
    return ''.join([str((int(ROTR_seventeen[i])+int(ROTR_eighteen[i])+int(SHR_ten[i]))%2) for i in range(len(num))])

def compute_intrm(H,K,W):
    a,b,c,d,e,f,g,h = H
    for t in range(64):
        T1 = binary_add([h,Sigma_one(e),choose(e,f,g),K[t],W[t]],32)
        T2 = binary_add([Sigma_zero(a),majority(a,b,c)],32)
        h=g
        g=f
        f=e
        e=binary_add([d,T1],32)
        d=c
        c=b
        b=a
        a=binary_add([T1,T2],32)
    return [H[0]+a,H[1]+b,H[2]+c,H[3]+d,H[4]+e,H[5]+f,H[6]+g,H[7]+h]

def Sigma_zero(num):
    ROTR_two = num[-2:]+num[:-2]
    ROTR_thirteen = num[-13:]+num[:-13]
    ROTR_twentytwo = num[-22:]+num[:-22]
    return ''.join([str((int(ROTR_two[i])+int(ROTR_thirteen[i])+int(ROTR_twentytwo[i]))%2) for i in range(len(num))])

def Sigma_one(num):
    ROTR_six = num[-6:]+num[:-6]
    ROTR_eleven = num[-11:]+num[:-11]
    ROTR_twentyfive = num[-25:]+num[:-25]
    return ''.join([str((int(ROTR_six[i])+int(ROTR_eleven[i])+int(ROTR_twentyfive[i]))%2) for i in range(len(num))])

def choose(num1,num2,num3):
    list = [num3,num2]
    return ''.join([str(list[int(num1[i:i+1])][i]) for i in range(len(num1))])

def majority(num1,num2,num3):
    temp = ''
    for i in range(len(num1)):
        if([num1[i],num2[i],num3[i]].count('0')>=2):
            temp += '0'
        else:
            temp += '1'
    return temp

main()
```
