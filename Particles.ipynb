```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
import math

particle_num = 50
steps = 10000
stepsize = .001
grav_const = .5
dot_size = 80
fig_size = 8
speed0 = 0

x = np.random.rand(particle_num)
y = np.random.rand(particle_num)
vx = speed0*np.random.rand(particle_num)
vy = speed0*np.random.rand(particle_num)
m0 = math.e*np.random.rand(particle_num)
m = dot_size*(np.exp(m0)/np.sum(np.exp(m0)))

fig, ax = plt.subplots(figsize=(fig_size,fig_size))
scatter = ax.scatter(x, y, s=dot_size*m)
plt.xlim(0,1)
plt.ylim(0,1)
ax.axis('off')
display(fig)

def distance(p1,p2):
    return math.sqrt((p1[0]-p2[0])**2+(p1[1]-p2[1])**2)

for step in range(steps):
    x += vx * stepsize
    y += vy * stepsize
    for i in range(particle_num):
        fx, fy = 0, 0
        ri = math.sqrt(dot_size*m[i]/math.pi/fig_size/100)
        for j in range(particle_num):
            if i != j:
                dist = distance((x[i], y[i]), (x[j], y[j]))
                rj = math.sqrt(m[j]/math.pi/fig_size/100)
                if dist > ri+rj:
                    force = grav_const * m[j] / (dist**2)
                    fx += (x[j] - x[i]) / dist * force
                    fy += (y[j] - y[i]) / dist * force
        vx[i] += fx * stepsize
        vy[i] += fy * stepsize

    scatter.set_offsets(np.c_[x, y])
    scatter.set_sizes(dot_size*m)
    ax.set_title(f"Step: {step}")
    clear_output(wait=True)
    display(fig)
    plt.pause(stepsize)
plt.show()
```
