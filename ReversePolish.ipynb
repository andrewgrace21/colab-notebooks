```python
# Logic
import math
import string
import sys

def getvars(list):
    vars = []
    for char in list:
        if char.isalpha() and char.isupper():
            if char not in vars and char!='V':
                vars.append(char)
    return vars

def evaluate(boolean,output):
    new = output[:]
    for i in range(len(new)):
        if new[i] in boolean:
            new[i] = boolean[new[i]]
    while len(new)>1:
        for i in range(len(new)):
            #print(i)
            #print(new)
            if i==len(new)-1:
                sys.exit('Invalid syntax')
            if not isinstance(new[i+1], bool):
                if new[i+1]=='!':
                    temp = not new[i]
                    #print(new[:i])
                    new = new[:i]+[temp]+new[i+2:]
                    break
                else:
                    sys.exit('Invalid syntax')
            elif not isinstance(new[i+2], bool) and (not new[i+2]=='!'):
                if new[i+2]=='^':
                    temp = new[i] and new[i+1]
                    new =  new[:i]+[temp]+new[i+3:]
                    break
                elif new[i+2]=='V':
                    temp = new[i] or new[i+1]
                    new = new[:i]+[temp]+new[i+3:]
                    break
                elif new[i+2]=='+':
                    temp = new[i] ^ new[i+1]
                    new = new[:i]+[temp]+new[i+3:]
                    break
                elif new[i+2]=='|':
                    temp = not (new[i] and new[i+1])
                    new = new[:i]+[temp]+new[i+3:]
                    break
                elif new[i+2]=='>':
                    temp = (False if (new[i]==True and new[i+1]==False) else True)
                    new = new[:i]+[temp]+new[i+3:]
                    break
                elif new[i+2]=='-':
                    temp = not (new[i] or new[i+1])
                    new = new[:i]+[temp]+new[i+3:]
                    break
                else:
                    sys.exit('Invalid syntax')
        #print(new)
    return ('T' if new[0]==True else 'F')

while True:
    text = input()
    initial = list(text)
    output = []
    stack = []

    value = {'^':1,'V':1,'!':2,'+':1,'|':1,'>':1,'-':1,'(':-math.inf,')':-math.inf}
    for l in string.ascii_uppercase:
        if l!='V':
            value[l] = 0

    for char in initial:
        if value[char]==0:
            output.append(char)
        elif char=='(':
            stack.append(char)
        elif char==')':
            while stack[-1]!='(':
                output.append(stack.pop())
            output.append(stack.pop())
            output.append(char)
        elif char=='!':
            stack.append(char)
        elif value[char]>value[stack[-1]]:
            stack.append(char)
        else:
            output.append(stack.pop())
            stack.append(char)
    while len(stack)>0:
        output.append(stack.pop())
    while '(' in output:
        output.remove('(')
    while ')' in output:
        output.remove(')')
    #print(output)

    vars = getvars(output)
    for j in range(len(vars)):
        print(f"  {vars[j]}  |",end='')
    print(f"  {text}  ")
    print((('-'*5)+'+')*len(vars)+'-'*(4+len(text)))
    for i in range(2**len(vars)):
        b = list(bin(i)[2:])
        if len(b)<len(vars):
            b = ['0']*(len(vars)-len(b))+b
        boolean = {vars[j]:(True if int(b[j])==0 else False) for j in range(len(b))}
        #print(b)
        sym = ['T' if int(b[j])==0 else 'F' for j in range(len(b))]
        for j in range(len(vars)):
            print(f"  {sym[j]}  |",end='')
        print(' '*(math.ceil(len(text)/2.0)+1),end='')
        print(evaluate(boolean,output))
        print(' '*(math.ceil(len(text)/2.0)+1))
#((PVQ)^(RVQ))
```
